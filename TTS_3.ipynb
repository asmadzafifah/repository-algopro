```python
# 1. Menyimpan data lima jenis komoditas pertanian ke dalam sebuah List
komoditas_list = ["Beras", "Cabai Merah", "Bawang Merah", "Jagung", "Kedelai"]
# 2. Menggunakan Dictionary untuk menyimpan detail salah satu komoditas
komoditas_detail = {
    "nama": "Cabai Merah",
    "harga_per_kg": 45000,
    "stok": 120
}
# 3. Menampilkan seluruh isi List awal
print("=== LIST KOMODITAS PERTANIAN (AWAL) ===")
print(komoditas_list)
# 4. Menambahkan satu komoditas baru ke dalam List menggunakan append()
komoditas_list.append("Minyak Goreng")
print("\n=== LIST KOMODITAS SETELAH DITAMBAHKAN ===")
print(komoditas_list)
# 5. Mengubah salah satu value pada Dictionary menggunakan key yang sesuai
# Mengubah harga cabai merah dari 45000 menjadi 50000
komoditas_detail["harga_per_kg"] = 50000
print("\n=== DETAIL KOMODITAS SETELAH DIPERBARUI ===")
print(f"Nama Komoditas : {komoditas_detail['nama']}")
print(f"Harga per kg   : Rp {komoditas_detail['harga_per_kg']}")
print(f"Stok           : {komoditas_detail['stok']} kg")
```

```python
# Definisi fungsi modular untuk menghitung total stok
def hitung_total_stok(stok_gudang):
    return sum(stok_gudang)
# Data stok dari dua kelompok gudang yang berbeda
stok_gudang_A = [150, 200, 350, 90]
stok_gudang_B = [500, 120, 230, 410, 60]
# Pemanggilan fungsi pertama
total_A = hitung_total_stok(stok_gudang_A)
# Pemanggilan fungsi kedua
total_B = hitung_total_stok(stok_gudang_B)
# Menampilkan hasil pemanggilan
print(f"Total stok Gudang Kelompok A: {total_A} kg")
print(f"Total stok Gudang Kelompok B: {total_B} kg")
```

**PERTANYAAN NOMOR 3**

```python
# 1. Fungsi untuk menyimpan data transaksi ke dalam Dictionary
def buat_transaksi(nama_produk, jumlah_terjual, harga_satuan):
    return {
        "nama_produk": nama_produk,
        "jumlah": jumlah_terjual,
        "harga_satuan": harga_satuan,
        "subtotal": jumlah_terjual * harga_satuan
    }
# 2. Fungsi untuk menghitung total pendapatan dari List berisi Dictionary
def hitung_total_pendapatan(daftar_transaksi):
    total = 0
    for transaksi in daftar_transaksi:
        total += transaksi["subtotal"]
    return total
# 3. Fungsi untuk menampilkan laporan akhir
def tampilkan_laporan(daftar_transaksi, total_pendapatan):
    print("=" * 45)
    print("        LAPORAN PENJUALAN PERTANI")
    print("=" * 45)
    print(f"{'Produk':<15} | {'Jumlah':<8} | {'Harga Satuan':<12}")
    print("-" * 45)
    for t in daftar_transaksi:
        print(f"{t['nama_produk']:<15} | {t['jumlah']:<8} | Rp {t['harga_satuan']:,}")
    print("-" * 45)
    print(f"TOTAL PENDAPATAN : Rp {total_pendapatan:,}")
    print("=" * 45)
# 4. Fungsi utama untuk mengatur alur program
def main():
    # Menyiapkan List untuk menampung seluruh data transaksi
    daftar_penjualan = []
    # Membuat beberapa data transaksi menggunakan fungsi 'buat_transaksi'
    t1 = buat_transaksi("Beras Super", 10, 15000)
    t2 = buat_transaksi("Cabai Merah", 5, 40000)
    t3 = buat_transaksi("Minyak Goreng", 12, 18000)
    # Menyimpan transaksi ke dalam List
    daftar_penjualan.append(t1)
    daftar_penjualan.append(t2)
    daftar_penjualan.append(t3)
    # Menghitung total pendapatan menggunakan fungsi 'hitung_total_pendapatan'
    total_pendapatan = hitung_total_pendapatan(daftar_penjualan)
    # Menampilkan laporan akhir menggunakan fungsi 'tampilkan_laporan'
    tampilkan_laporan(daftar_penjualan, total_pendapatan)
# Menjalankan fungsi utama
if __name__ == "__main__":
    main()
```

**PERTANYAAN NOMOR 4**

```python
# 1. Fungsi untuk mengelola daftar nama petani (List)
def kelola_nama_petani():
    petani = ["Budi", "Siti", "Joko"]
    petani.append("Agus")  # Menambahkan petani baru
    return petani
# 2. Fungsi untuk menyimpan identitas tetap petani (Tuple)
def buat_identitas_petani():
    # Format: (Kode Anggota, Tanggal Bergabung)
    identitas = ("KOP-2024-001", "15-01-2024")
    return identitas
# 3. Fungsi untuk mencatat detail transaksi (Dictionary)
def buat_detail_transaksi():
    transaksi = {
        "id_transaksi": "TRX-001",
        "komoditas": "Beras Super",
        "jumlah_kg": 50,
        "harga_per_kg": 14000
    }
    return transaksi
# 4. Fungsi untuk mengumpulkan jenis komoditas unik (Set)
def kelola_komoditas_unik():
    # Memasukkan beberapa komoditas yang menyertakan duplikat
    komoditas_input = ["Beras", "Cabai", "Beras", "Jagung", "Cabai", "Kedelai"]
    komoditas_unik = set(komoditas_input)  # Otomatis menyaring duplikat
    return komoditas_unik
# 5. Fungsi utama untuk mengatur seluruh alur sistem
def main():
    # Memanggil fungsi-fungsi modular
    daftar_petani = kelola_nama_petani()
    identitas_petani = buat_identitas_petani()
    transaksi = buat_detail_transaksi()
    komoditas_unik = kelola_komoditas_unik()
    # Menampilkan Laporan Sistem Koperasi
    print("=" * 50)
    print("      SISTEM PENCATATAN KOPERASI TANI")
    print("=" * 50)
    print("\n1. Identitas Anggota (Tuple):")
    print(f"   Kode Anggota     : {identitas_petani[0]}")
    print(f"   Tanggal Bergabung : {identitas_petani[1]}")
    print("\n2. Daftar Nama Petani Aktif (List):")
    for idx, nama in enumerate(daftar_petani, start=1):
        print(f"   {idx}. {nama}")
    print("\n3. Detail Transaksi Terakhir (Dictionary):")
    print(f"   ID Transaksi : {transaksi['id_transaksi']}")
    print(f"   Komoditas    : {transaksi['komoditas']}")
    print(f"   Jumlah       : {transaksi['jumlah_kg']} kg")
    print(f"   Total Harga  : Rp {transaksi['jumlah_kg'] * transaksi['harga_per_kg']:,}")
    print("\n4. Daftar Jenis Komoditas Unik (Set):")
    print(f"   {', '.join(komoditas_unik)}")
    print("=" * 50)
# Menjalankan program utama
if __name__ == "__main__":
    main()
```

**PERTANYAAN NOMOR 5**

```python
import unittest
# ==============================================================================
# 1. VARIABLE GLOBAL
# ==============================================================================
# Variabel global yang dapat diakses oleh fungsi di dalam modul ini
PAJAK_GUDANG = 0.05  # Pajak penyimpanan 5%
# ==============================================================================
# 2. FUNGSI MODULAR & PENGGUNAAN STRUKTUR DATA
# ==============================================================================

def buat_data_barang(nama, stok, harga_per_kg, lokasi_tuple):
    """
    Fungsi 1: Membuat Dictionary data barang.
    Struktur Data:
    - Dictionary : Menampung atribut barang (key-value).
    - Tuple      : Mengunci data lokasi gudang agar immutable, misal ("Blok-A", "Rak-1").
    """
    # Variable Local: hanya berlaku di dalam fungsi ini
    nama_formatted = nama.strip().title()
    return {
        "nama": nama_formatted,
        "stok": stok,
        "harga_per_kg": harga_per_kg,
        "lokasi": lokasi_tuple  # Menggunakan Tuple
    }
def hitung_nilai_total(daftar_barang):
    """
    Fungsi 2: Menghitung total nilai inventaris.
    Struktur Data:
    - List : Paramenter 'daftar_barang' berupa List berisi Dictionary.
    """
    if not daftar_barang:
        return 0.0
    # Variable Local
    subtotal = sum(barang["stok"] * barang["harga_per_kg"] for barang in daftar_barang)
    # Mengakses Variable Global 'PAJAK_GUDANG'
    total_akhir = subtotal + (subtotal * PAJAK_GUDANG)
    return round(total_akhir, 2)
def cek_peringatan_stok(daftar_barang, stok_minimal=10):
    """
    Fungsi 3: Mengecek barang yang stoknya kurang dari atau sama dengan threshold.
    Fitur Python:
    - Default Argument : 'stok_minimal=10' sebagai nilai default jika tidak diisi.
    """
    # Variable Local
    barang_kritis = []
    for barang in daftar_barang:
        if barang["stok"] <= stok_minimal:
            barang_kritis.append(barang["nama"])
    return barang_kritis
def ekstrak_kategori_unik(daftar_kategori):
    """
    Fungsi 4: Mengolah input list kategori menjadi himpunan unik.
    Struktur Data:
    - Set : Mengeliminasi duplikasi data kategori secara otomatis.
    """
    return set(kategori.strip().lower() for kategori in daftar_kategori)
def tampilkan_laporan(daftar_barang, kategori_set, total_nilai):
    """
    Fungsi 5: Menampilkan output laporan inventaris.
    """
    print("=" * 55)
    print("        LAPORAN INVENTARIS HASIL PANEN TANI")
    print("=" * 55)
    print(f"Kategori Unik (Set) : {', '.join(kategori_set).title()}")
    print("-" * 55)
    print(f"{'Nama Barang':<18} | {'Stok':<8} | {'Harga/kg':<10} | {'Lokasi (Tuple)':<10}")
    print("-" * 55)
    for b in daftar_barang:
        lokasi_str = f"{b['lokasi'][0]}-{b['lokasi'][1]}"
        print(f"{b['nama']:<18} | {b['stok']:<8} | Rp{b['harga_per_kg']:<8} | {lokasi_str:<10}")
    print("-" * 55)
    print(f"Total Nilai (+Pajak Global 5%): Rp {total_nilai:,.2f}")
    print("=" * 55)

# ==============================================================================
# 3. FUNGSI UTAMA (MAIN)
# ==============================================================================

def main():
    print("\n--- JALANKAN PROGRAM UTAMA ---")
    # 1. List: Menyimpan daftar barang
    inventaris_list = []
    # 2. Tuple: Lokasi gudang (Blok, Rak)
    lokasi_1 = ("Blok-A", "Rak-1")
    lokasi_2 = ("Blok-B", "Rak-3")
    lokasi_3 = ("Blok-A", "Rak-2")
    # 3. Dictionary & List
    b1 = buat_data_barang("Beras Pandan Wangi", 100, 15000, lokasi_1)
    b2 = buat_data_barang("Cabai Merah", 5, 40000, lokasi_2)
    b3 = buat_data_barang("Jagung Manis", 0, 8000, lokasi_3) # Edge case stok 0
    inventaris_list.extend([b1, b2, b3])
    # 4. Set: Mengolah data kategori dengan duplikat
    raw_kategori = ["Beras", "Cabai", "Beras", "Jagung", "Cabai"]
    kategori_unik = ekstrak_kategori_unik(raw_kategori)
    # Perhitungan Total
    total_nilai = hitung_nilai_total(inventaris_list)
    # Tampilkan Laporan Akhir
    tampilkan_laporan(inventaris_list, kategori_unik, total_nilai)
    # Penggunaan Default Argument
    peringatan_default = cek_peringatan_stok(inventaris_list) # Menggunakan default (<=10)
    print(f"\n[PERINGATAN] Stok Kritis (<= 10 kg) : {peringatan_default}")

# ==============================================================================
# 4. PENGUJIANSederhana (ASSERT)
# ==============================================================================

def jalankan_pengujian_assert():
    print("\n--- MENJALANKAN PENGUJIAN ASSERT ---")
    # Uji 1: hitung_nilai_total (Edge Case List Kosong)
    assert hitung_nilai_total([]) == 0.0, "Gagal: List kosong harus menghasilkan 0.0"
    # Uji 2: hitung_nilai_total (Kalkulasi Normal + Pajak 5%)
    # Subtotal = 10 * 10.000 = 100.000 -> Total + Pajak = 105.000
    sample_data = [{"nama": "Padi", "stok": 10, "harga_per_kg": 10000}]
    assert hitung_nilai_total(sample_data) == 105000.0, "Gagal: Perhitungan pajak salah"
    # Uji 3: cek_peringatan_stok (Default Argument & Custom Argument)
    data_stok = [{"nama": "Beras", "stok": 15}, {"nama": "Cabai", "stok": 5}]
    assert cek_peringatan_stok(data_stok) == ["Cabai"], "Gagal: Default argument stok minimal salah"
    assert cek_peringatan_stok(data_stok, stok_minimal=20) == ["Beras", "Cabai"], "Gagal: Custom argument stok minimal salah"
    # Uji 4: ekstrak_kategori_unik (Pembersihan Duplikat & Format String)
    kat_input = ["Beras", "BERAS ", "cabai"]
    assert ekstrak_kategori_unik(kat_input) == {"beras", "cabai"}, "Gagal: Pengolahan Set gagal"
    print("Result: Semua pengujian assert BERHASIL (PASSED)!")

# ==============================================================================
# 5. PENGUJIAN TERSTRUKTUR (UNITTEST)
# ==============================================================================

class TestInventaris(unittest.TestCase):
    def test_hitung_nilai_total_normal(self):
        # Subtotal = (10*10000) = 100.000 + Pajak 5% = 105.000
        sample = [{"nama": "Jagung", "stok": 10, "harga_per_kg": 10000}]
        self.assertEqual(hitung_nilai_total(sample), 105000.0)
    def test_hitung_nilai_total_list_kosong(self):
        # Edge case: input list kosong
        self.assertEqual(hitung_nilai_total([]), 0.0)
    def test_cek_peringatan_stok_default(self):
        # Menguji default argument stok_minimal = 10
        sample = [{"nama": "Beras", "stok": 12}, {"nama": "Cabai", "stok": 10}]
        self.assertEqual(cek_peringatan_stok(sample), ["Cabai"])
    def test_ekstrak_kategori_unik(self):
        sample = ["Padi", "PADI", " Padi "]
        self.assertEqual(ekstrak_kategori_unik(sample), {"padi"})

# ==============================================================================
# EKSKUSI PROGRAM
# ==============================================================================

if __name__ == "__main__":
    # 1. Jalankan Fungsi Utama
    main()
    # 2. Jalankan Pengujian Assert
    jalankan_pengujian_assert()
    # 3. Jalankan Pengujian Unittest
    print("\n--- MENJALANKAN PENGUJIAN UNITTEST ---")
    unittest.main(exit=False)
```
